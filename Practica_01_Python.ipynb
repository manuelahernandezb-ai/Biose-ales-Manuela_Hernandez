REPRESENTACIÓN DE SEÑALES DISCRETAS

Impulso:

```python
def impseq(n0,n1,n2):
    # Genera x(n) = delta(n-n0); n1 <= n <= n2
    n=np.arange(n1,n2+1)
    x=((n-n0)==0).astype(int)#Sirve para convertir los valores booleanos (True y False) que genera la comparación lógica en números enteros (1 y 0). También puedo usar float.
    return[x,n]

x_imp, n_imp = impseq(0, -5, 5) #Aqui estoy llamando la función impseq, vero ya definí sus argumentos para el ejemplo.
plt.figure(figsize=(6, 3))
plt.stem(n_imp, x_imp)
plt.title("Impulso unitario δ(n)")
plt.xlabel("n")
plt.ylabel("Amplitud")
plt.grid(True)
plt.show()
```

Escalon unitario:

```python
def escuni(n0,n1,n2):
    # Genera x(n) = u(n-n0); n1 <= n <= n2
    n=np.arange(n1,n2+1)
    x=((n-n0)>=0).astype(int)
    return[x,n]

x_u, n_u = escuni(0, -5, 10)
plt.figure(figsize=(6, 3))
plt.stem(n_u, x_u)
plt.title("Escalón unitario u(n)")
plt.xlabel("n")
plt.ylabel("Amplitud")
plt.grid(True)
plt.show()
```

Rampa:

```python
def rampa(n0,n1,n2,m=1):
    # Genera x(n) = m*(n-n0) para n>=n0, 0 en otro caso; n1 <= n <= n2
    n=np.arange(n1,n2+1)
    x = np.where(n >= n0, m * (n - n0), 0.0)
    return [x,n]


x_r, n_r = rampa(0, -5, 10, m=1)
plt.figure(figsize=(6, 3))
plt.stem(n_r, x_r)
plt.title("Rampa r(n), pendiente m=1")
plt.xlabel("n")
plt.ylabel("Amplitud")
plt.grid(True)
plt.show()
```

R.Genere las siguientes secuencias usando las funciones básicas de Python que se han presentado. Grafique los resultados.



A.𝑥1(𝑛)= 3𝛿(𝑛+1)+5𝛿(𝑛+3)+3𝛿(𝑛+2)+3𝛿(𝑛)+ 𝛿(𝑛). ¿Cómo debe ser el vector de muestras?

```python
def impulso(n0,n1,n2):
    n=np.arange(n1,n2+1) #El vector de muestras n debe estar compuesto por números enteros contiguos con un paso de 1 (np.arange(-5, 5)), cubriendo obligatoriamente el intervalo de $n = -3$ a $n = 0$ donde se encuentran los impulsos.
    x=((n-n0)==0).astype(int)
    return[x,n]

#Definir el vector de muestras
n1=-5 
n2=5

#Generar cada impulso
x_n3, n= impulso(-3,n1,n2) # delta(n + 3) - n0 = -3
x_n2, n= impulso(-2,n1,n2) # delta(n + 2) - n0 = -2
x_n1, n= impulso(-1,n1,n2) # delta(n + 1) - n0 = -1
x_n0, n= impulso(0,n1,n2) # delta(n) - n0 = -0

#Combinar linealmente los impulsos
x1= 5 * x_n3 + 3 * x_n2 + 3 + x_n1 + 3 + x_n0 + 1* x_n0

plt.figure(figsize=(7,3.5))
plt.stem(n,x1)
plt.title ("Sinusoides")
plt.xlabel('n')
plt.ylabel('Amplitud')
plt.grid(True)
plt.show()
```
