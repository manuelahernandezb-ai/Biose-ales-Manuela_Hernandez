# Punto 1. Revisión teórica
Del artículo https://www.nature.com/articles/s41598-020-59821-7, hacer un resumen de la sección extracción de características (Features extraction) y discutir desde el artículo cómo se hace cuando desaparecen ciertas formas de ondas en el complejo PQRS debido a alguna enfermedad (10%)

### Resumen de la sección de extracción de características
Los autores parten de que hay dos formas comunes de obtener características de un ECG. La primera es usar redes neuronales profundas, que aprenden por sí solas transformaciones de la señal cruda. Sin embargo, estas requieren muchos datos y tiempo de cómputo, y sus características no se pueden interpretar. La segunda es medir directamente las ondas P, Q, T y el complejo QRS, con sus amplitudes, duraciones y distancias entre picos. El problema es que estas medidas no alcanzan para clasificar con alta precisión arritmias en las que algunas ondas se deforman o desaparecen. Además, los autores señalan que extraer solo características en frecuencia, con la transformada de Fourier o wavelet, hace perder la información temporal de la señal.

Por otro lado, la forma de las ondas cambia entre personas y en una misma persona, y también depende de la edad, el género y la raza. Por eso, antes de extraer características, los autores normalizan cada señal con el método de máximo y mínimo para que todas queden en la misma escala de amplitud. También agregan la edad y el género como características.

Con base en esto, proponen un método nuevo e interpretable que organizan en 11 grupos de características, desde 11 hasta 39.830 variables. El grupo 1 contiene las 11 medidas que entrega el propio electrocardiógrafo, como la frecuencia ventricular y auricular, la duración del QRS, el intervalo QT, los ejes R y T, el número de complejos QRS y el inicio y fin de la onda Q. El grupo 2 se calcula en la derivación II e incluye la media, la varianza y el número de intervalos RR, junto con la media y la varianza de la altura, el ancho y la prominencia de los picos y valles de la señal.

Para obtener estas medidas, los autores detectan todos los máximos y mínimos locales de la señal y los dividen en tres conjuntos, que son los complejos QRS, los picos que no son QRS y los valles. A cada uno le miden la altura, el ancho y la prominencia. La prominencia indica cuánto sobresale un pico respecto a los picos vecinos, y se define como la distancia vertical entre el pico y la línea de contorno más baja que lo rodea.

El aporte principal del artículo es el grupo 6, también calculado en la derivación II. Para cada par de conjuntos, por ejemplo QRS contra QRS o QRS contra valles, se calculan las diferencias de posición entre ellos y las razones entre la diferencia de altura, de ancho o de prominencia y la diferencia de tiempo. Como cada paciente tiene un número distinto de picos, estos valores no se pueden usar directamente en un clasificador. Por eso, los autores construyen una distribución de frecuencias empírica, es decir, un histograma de 100 intervalos entre el valor mínimo y el máximo de cada medida. Así, cada variable queda representada siempre por 100 números, sin importar cuántos latidos tenga el registro. En total, el grupo 6 tiene 3.302 características, y los demás grupos combinan los grupos 1, 2 y 6, en la derivación II o en las 12 derivaciones.

### Discusión sobre las ondas que desaparecen
El artículo reconoce que algunas arritmias deforman o eliminan ondas del complejo PQRS. Por ejemplo, en la fibrilación auricular y el aleteo auricular la onda P se reemplaza por varias ondas de fibrilación o aleteo, más pequeñas y sin relación con el ritmo del QRS. En estos casos, un método que busca la onda P en una posición fija falla, porque la onda no existe o aparece en un lugar distinto.

Para resolverlo, los autores no intentan identificar cada onda por su nombre. En lugar de eso, detectan todos los picos y valles de la señal y los clasifican solo en complejos QRS, picos que no son QRS y valles. De esta forma, si la onda P desaparece, simplemente cambia la cantidad y la forma de los picos que no son QRS, y ese cambio queda registrado en sus histogramas. En la fibrilación auricular, por ejemplo, aparecen muchos picos pequeños e irregulares, lo que modifica la distribución de alturas, anchos y prominencias. Además, como el histograma siempre tiene 100 valores, el número de características no cambia aunque falten ondas.

Asimismo, las razones entre las diferencias de altura, ancho o prominencia y la diferencia de tiempo describen cómo se relacionan los picos a lo largo del registro. Según los autores, esta relación entre la forma de las ondas y su duración es clave para reconocer el ritmo. Por último, las características del intervalo RR, como su media, su varianza y su número, no dependen de la onda P, ya que se calculan con los complejos QRS, que casi siempre están presentes. Por eso permiten detectar ritmos irregulares como la fibrilación auricular aunque la onda P no se vea.
